# Fully Connected Feedforward Network

Características principales: En esta arquitectura, la información fluye en una sola dirección (hacia adelante) desde la entrada hasta la salida, y no hay conexiones hacia atrás (también conocidas como bucles o ciclos). Además, se llama "totalmente conectada" porque cada neurona de una capa está conectada a todas las neuronas de la capa siguiente.

## Teoría y Práctica: Retropropagación (Backpropagation) en Redes Neuronales

Este documento explica la teoría del **Capítulo 3: Sigmoid Neurons and Backpropagation** mediante un ejemplo numérico práctico, cubriendo desde el cálculo hacia adelante (Forward Pass) hasta la actualización de pesos usando el Descenso de Gradiente.

### 1. El Planteamiento Teórico y las Funciones de Activación

Para que una red neuronal aprenda, usamos el algoritmo de **Descenso de Gradiente**, el cual requiere calcular derivadas (qué tanto afecta un pequeño cambio de un peso al error final). La antigua función de "signo" no servía porque no se puede derivar en todos sus puntos (tiene un salto abrupto). 

Por ello, usamos funciones en forma de "S" (diferenciables en todo punto):
* **Capa Oculta (Neurona G):** Usa la **Tangente Hiperbólica (tanh)**. Su rango es de -1 a 1, ideal para mantener los datos centrados en cero.
* **Capa de Salida (Neurona F):** Usa la **Sigmoide Logística ($S(x)$)**. Su rango es de 0 a 1, ideal para interpretar la salida como una probabilidad.

### Datos Iniciales (Ejemplo de Entrenamiento)

| Componente | Símbolo | Valor Inicial |
| :--- | :--- | :--- |
| **Entrada 1** | $x_1$ | -0.9 |
| **Entrada 2** | $x_2$ | 0.1 |
| **Valor Esperado** | $y_{truth}$ | 1.0 |
| **Tasa de Aprendizaje** | $lr$ | 0.1 |
| **Pesos a Neurona G (Oculta)** | $w_{xg0}$ (sesgo) | 0.3 |
| | $w_{xg1}$ | 0.6 |
| | $w_{xg2}$ | -0.1 |
| **Pesos a Neurona F (Salida)**| $w_{gf0}$ (sesgo) | -0.2 |
| | $w_{gf1}$ | 0.5 |

---

## 2. El Paso Hacia Adelante (Forward Pass)

En esta fase, los datos fluyen desde la entrada hasta la salida. Multiplicamos las entradas por los pesos, sumamos el sesgo y pasamos el resultado por la función de activación.

**A. Cálculo de la Neurona G (Capa Oculta):**
$$y_g = \tanh(w_{xg0} + w_{xg1}x_1 + w_{xg2}x_2)$$
$$y_g = \tanh(0.3 + 0.6(-0.9) + (-0.1)(0.1)) = \tanh(-0.25) = -0.24$$

**B. Cálculo de la Neurona F (Capa de Salida):**
$$y_f = S(w_{gf0} + w_{gf1}y_g)$$
$$y_f = S(-0.2 + 0.5(-0.24)) = S(-0.32) = 0.42$$

**C. Cálculo del Error (Loss Function):**
Calculamos qué tan equivocada está la red usando el Error Cuadrático Medio (MSE). Se divide entre 2 para simplificar la derivada más adelante.
$$MSE = \frac{(y_{truth} - y_f)^2}{2} = \frac{(1.0 - 0.42)^2}{2} \approx 0.17$$

---

## 3. La Regla de la Cadena y el Paso Hacia Atrás (Backward Pass)

Para corregir el error de 0.17, necesitamos propagar la "culpa" hacia atrás usando la **Regla de la Cadena**. Esto nos evita recalcular combinaciones desde cero.

**A. Derivada del Error Final:**
$$MSE' = -(y_{truth} - y_f) = -(1.0 - 0.42) = -0.58$$

**B. Término de Error para la Neurona F:**
Multiplicamos el error final por la derivada de la función sigmoide, que es $S(1-S)$.
$$Error_{f} = MSE' \cdot y_f(1 - y_f)$$
$$Error_{f} = -0.58 \cdot 0.42(1 - 0.42) = -0.14$$

**C. Término de Error para la Neurona G:**
Propagamos el error de la salida hacia la capa oculta. Multiplicamos el error de F por el peso que las conecta ($w_{gf1}$) y por la derivada de la tangente hiperbólica, que es $1 - \tanh^2(x)$.
$$Error_{g} = Error_{f} \cdot w_{gf1} \cdot (1 - y_g^2)$$
$$Error_{g} = -0.14 \cdot 0.5 \cdot (1 - (-0.24)^2) \approx -0.066$$

---

## 4. Actualización de Pesos (Weight Adjustment)

Calculamos cuánto cambiará cada peso ($\Delta w$). La regla es:
$$\Delta w = -lr \cdot \text{entrada} \cdot \text{Error de la neurona}$$

**Para la Neurona G:**
* $\Delta w_{xg0}$ (sesgo, entrada = 1): $$-0.1 \cdot 1 \cdot (-0.066) = 0.0066$$
* $\Delta w_{xg1}$ (entrada = $x_1$): $$-0.1 \cdot (-0.9) \cdot (-0.066) = -0.0060$$
* $\Delta w_{xg2}$ (entrada = $x_2$): $$-0.1 \cdot 0.1 \cdot (-0.066) = 0.00066$$

**Para la Neurona F:**
* $\Delta w_{gf0}$ (sesgo, entrada = 1): $$-0.1 \cdot 1 \cdot (-0.14) = 0.014$$
* $\Delta w_{gf1}$ (entrada = $y_g$): $$-0.1 \cdot (-0.24) \cdot (-0.14) = -0.0035$$

**Nuevos Pesos Finales:** (Sumando $\Delta w$ al peso original)
* $w_{xg0_{nuevo}} = 0.3 + 0.0066 = 0.3066$
* $w_{xg1_{nuevo}} = 0.6 - 0.0060 = 0.5940$
* $w_{xg2_{nuevo}} = -0.1 + 0.00066 = -0.09934$
* $w_{gf0_{nuevo}} = -0.2 + 0.014 = -0.186$
* $w_{gf1_{nuevo}} = 0.5 - 0.0035 = 0.4965$

---

## 5. Conclusión de Arquitecturas

Este algoritmo base funciona para redes de múltiples tamaños:
1. **Redes Totalmente Conectadas (Feedforward):** Como el ejemplo anterior, la información fluye en una sola dirección. Si la capa oculta tuviera más neuronas, el término de error simplemente sumaría las "culpas" ponderadas de todas las neuronas a las que alimenta.
2. **Redes Convolucionales (CNN):** Funcionan igual, pero comparten pesos entre varias neuronas para ser más eficientes (ideal para imágenes).
3. **Redes Recurrentes (RNN):** Permiten ciclos hacia atrás para recordar información pasada, modificando ligeramente cómo fluyen los datos, pero operando bajo los mismos principios fundamentales de propagación de gradientes.

# El Truco Matemático del Error Cuadrático Medio (MSE)

## 1. ¿Por qué es "legal" alterar la fórmula original?
[cite_start]El objetivo principal del algoritmo de entrenamiento es encontrar los valores de las variables (los pesos) que logren minimizar la función de error[cite: 329]. 

[cite_start]En matemáticas de optimización, el punto mínimo de una función no cambia si divides toda la función por un número[cite: 329]. [cite_start]Es decir, los valores de los pesos que minimizan una expresión como $(y - f)^2$ son exactamente los mismos valores que minimizan la expresión $\frac{(y - f)^2}{2}$[cite: 329]. [cite_start]Por lo tanto, dividir la función entre una constante es una maniobra matemáticamente válida que no altera el resultado final de nuestro entrenamiento[cite: 329].

## 2. ¿Por qué resulta útil? (El truco de cálculo diferencial)
[cite_start]El autor agregó ese divisor de 2 específicamente porque simplificará las matemáticas más adelante durante el paso hacia atrás (retropropagación)[cite: 328].

[cite_start]Cuando aplicamos la retropropagación, el primer paso es calcular la derivada parcial de la función de error con respecto a la predicción de la red (la salida $f$)[cite: 363, 364].

Si usamos la porción del error al cuadrado "normal" para un solo ejemplo, la función base es:
$$E = (y - f)^2$$

Si derivamos esto usando la Regla de la Cadena, el exponente $2$ baja a multiplicar, y luego multiplicamos por la derivada de lo que está adentro del paréntesis (que es $-1$ debido al signo negativo de $f$):
$$E' = 2 \cdot (y - f) \cdot (-1) = -2(y - f)$$

[cite_start]Para evitar arrastrar ese multiplicador $2$ en todos los cálculos posteriores en la red, el autor redefine la función de error ($e$) desde el principio añadiendo el divisor[cite: 326, 328]:
[cite_start]$$e(f) = \frac{(y - f)^2}{2}$$ [cite: 326]

[cite_start]Ahora, observa lo que ocurre al derivar esta versión modificada usando las reglas del cálculo[cite: 364]:
[cite_start]$$\frac{\partial e}{\partial f} = \frac{2(y - f)}{2} \cdot (-1)$$ [cite: 364]

[cite_start]¡Los números 2 se cancelan matemáticamente![cite: 364]. [cite_start]El $2$ que bajó del exponente se anula perfectamente con el $2$ que colocamos en el denominador de la fórmula[cite: 364]. El resultado queda mucho más limpio y fácil de programar:
[cite_start]$$\frac{\partial e}{\partial f} = -(y - f)$$ [cite: 364]

---
[cite_start]**En resumen:** Se divide la fórmula original entre 2 como una conveniencia matemática previa[cite: 328]. [cite_start]Así, al momento de aplicar derivadas, los números se cancelan solos y la ecuación final de ajuste de pesos requiere menos multiplicaciones[cite: 328, 364].
