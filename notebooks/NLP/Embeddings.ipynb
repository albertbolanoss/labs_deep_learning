# Apuntes sobre Word2Viz, Embeddings y Distancia Vectorial

## 1. ¿Qué hay detrás de "word2viz"? (Word Vectors / Embeddings)

Los modelos de IA no entienden texto; entienden números. Algoritmos como Word2Vec o GloVe toman enormes cantidades de texto y convierten cada palabra en un arreglo numérico (un vector) de alta dimensionalidad. 

Lo fascinante de estos vectores es que capturan el significado semántico de las palabras basándose en su contexto. Las matemáticas literales se pueden aplicar a los conceptos. La ecuación clásica en NLP es:

$$\vec{Rey} - \vec{Hombre} + \vec{Mujer} \approx \vec{Reina}$$

El problema es que un vector puede tener cientos de dimensiones, algo imposible de visualizar para el cerebro humano. 

### ¿Qué es word2viz?
**Word2viz** es una interfaz visual que toma esos vectores de alta dimensionalidad y los proyecta en un plano 2D que podemos entender, utilizando ejes personalizados.

Por ejemplo, al mapear capitales y países:
*   **Eje Y (`city` a `country`):** Separa matemáticamente los conceptos de ciudades y países.
*   **Eje X (`west` a `east`):** Organiza la geografía. 
*   **El resultado:** Las relaciones entre París-Francia, Londres-Inglaterra y Beijing-China forman líneas casi paralelas. El algoritmo aprendió la relación "capital-país" únicamente analizando la proximidad de estas palabras en millones de textos.

### El problema del Sesgo (Bias)
Debido a que estos modelos se entrenan con datos creados por humanos, absorben e internalizan todos los prejuicios y estereotipos presentes en esos textos. Si cambias la vista en word2viz a "Trabajos" (Jobs), verás que ciertas profesiones se agrupan matemáticamente más cerca del género masculino o femenino. El modelo ha aprendido un sesgo de género.

### Uso Práctico en la Ingeniería y Arquitectura
*   **Auditoría de Datos:** Word2viz sirve como una herramienta de depuración (*debugging*) visual para detectar sesgos.
*   **Diseño de Pipelines Seguros:** Inyectar *embeddings* sesgados en sistemas de búsqueda corporativos o arquitecturas multi-agente puede generar resultados discriminatorios.
*   **Mitigación:** La práctica segura es asumir que tu modelo tiene sesgos ocultos e implementar pruebas exhaustivas en los entornos de validación antes de llevar el sistema a producción.

### ¿Qué representa cada dimensión? (El Espacio Latente)
En los modelos de deep learning, las dimensiones no son explícitas ni interpretables para los humanos. El algoritmo no tiene un diccionario de conceptos aislados (como "disciplina" o "estado físico"). En su lugar, esos conceptos están **distribuidos** a través de una combinación matemática compleja de todas las dimensiones. Cada dimensión representa un "eje de variación semántica" abstracto que la máquina encontró útil para diferenciar palabras.

### ¿Por qué arrays multidimensionales?
El lenguaje humano es complejo y necesita separar palabras en función de género, tiempo verbal, tono, categoría gramatical y contexto semántico. Al tener miles de dimensiones, el algoritmo tiene suficiente "espacio" para separar los matices semánticos y ubicar cada concepto exactamente donde pertenece.

### Modelo Mental para comprender el algoritmo
1.  **Imagina una galaxia:** Visualiza los embeddings como estrellas en un universo de miles de dimensiones.
2.  **El significado es la ubicación:** Una palabra se define por **dónde** está ubicada en esa galaxia.
3.  **La distancia es la similitud:** Los algoritmos toman tu pregunta, la convierten en un punto en esa galaxia, y buscan los documentos o palabras que estén físicamente más cerca de esa coordenada.

## 3. ¿Cómo se calcula la distancia entre 2 vectores?

Para calcular la distancia (o similitud) entre dos vectores (por ejemplo, el vector $A$ y el vector $B$, ambos con $n$ dimensiones), se comparan los arrays dimensión por dimensión.

### 1. Similitud del Coseno (Cosine Similarity)
Mide el **ángulo** entre los dos vectores. Ignora la "magnitud" (el tamaño) del vector. Es el método más utilizado en NLP.

$$ \text{Similitud}(A, B) = \cos(\theta) = \frac{A \cdot B}{\|A\| \|B\|} = \frac{\sum_{i=1}^{n} A_i B_i}{\sqrt{\sum_{i=1}^{n} A_i^2} \sqrt{\sum_{i=1}^{n} B_i^2}} $$

*   **1:** Apuntan en la misma dirección (significado idéntico).
*   **0:** Son ortogonales (sin relación semántica aparente).
*   **-1:** Apuntan en direcciones opuestas (antónimos perfectos).

### 2. Distancia Euclidiana (L2 Distance)
Mide físicamente qué tan separados están los dos vectores en línea recta dentro del espacio multidimensional.

$$ d(A, B) = \sqrt{\sum_{i=1}^{n} (A_i - B_i)^2} $$

*   Es un número positivo. Mientras más cercano a **0**, más similares son los vectores.

### 3. Producto Punto (Dot Product)
Es el método más rápido a nivel computacional, muy utilizado en bases de datos vectoriales cuando los vectores ya han sido normalizados (longitud 1).

$$ A \cdot B = \sum_{i=1}^{n} A_i B_i $$

*   Si los vectores están normalizados, da exactamente el mismo resultado que la Similitud del Coseno, pero requiere menos operaciones.

### Resumen Comparativo

| Método | ¿Qué mide principalmente? | Rango típico | Uso recomendado |
| :--- | :--- | :--- | :--- |
| **Similitud de Coseno** | Ángulo (dirección semántica) | -1 a 1 | Motores de búsqueda, comparación de textos, sistemas de recomendación. |
| **Distancia Euclidiana** | Distancia física (magnitud y ángulo) | 0 a infinito | Agrupación (*Clustering* como K-Means), detección de anomalías. |
| **Producto Punto** | Proyección de un vector sobre otro | -infinito a infinito | Cálculos de alto rendimiento en bases de datos vectoriales (con vectores normalizados). |

# La Matemática de la Similitud y el Producto Punto

## 1. La Matemática de la Similitud del Coseno

La fórmula de la similitud del coseno mide el ángulo entre dos vectores y se define como:

$$ \cos(\theta) = \frac{A \cdot B}{\|A\| \|B\|} $$

Esta ecuación tiene dos partes fundamentales que trabajan en conjunto: el numerador y el denominador.

### El Numerador: $A \cdot B$ (El Producto Punto)
Esta parte calcula qué tanto coinciden las características de ambos vectores, multiplicando sus dimensiones una por una y sumando los resultados.
*   **Fórmula:** $A_1 B_1 + A_2 B_2 + \dots + A_n B_n$
*   Si ambos vectores tienen un valor numérico alto en la misma dimensión exacta, la multiplicación dará un número positivo grande, empujando la puntuación final hacia arriba.

### El Denominador: $\|A\| \|B\|$ (La Normalización)
Aquí calculamos la "longitud" o "magnitud" física de cada vector de forma independiente (usando una extensión del teorema de Pitágoras) y luego multiplicamos esas dos longitudes.
*   **Fórmula de la magnitud:** $\|A\| = \sqrt{A_1^2 + A_2^2 + \dots + A_n^2}$
*   **¿Para qué sirve?** El producto punto por sí solo puede dar números gigantescos (por ejemplo, **5000**) si estás analizando un documento muy largo lleno de palabras repetidas. Al dividir el producto punto entre la longitud geométrica de los vectores, "escalamos" o "penalizamos" el resultado para que siempre caiga en un rango estandarizado entre **-1 y 1**. Esto elimina el sesgo del tamaño, permitiéndote comparar limpiamente un documento de 50 páginas con una consulta de búsqueda de una sola oración.

## 2. ¿Por qué el Producto Punto sirve para hallar la similitud?

Es importante hacer una distinción técnica inicial: el producto punto no mide exactamente una "distancia" geométrica (donde **0** significaría que son idénticos), sino la **similitud o alineación** (donde un número mayor significa que son más afines).

Imagina que tenemos vectores simplificados de 2 dimensiones que representan los conceptos de `[Tecnología, Finanzas]`.
*   **Vector A** (Artículo sobre microservicios y contenedores): `[0.9, 0.1]`
*   **Vector B** (Documentación de Spring Boot): `[0.8, 0.0]`
*   **Vector C** (Reporte de ingresos corporativos): `[0.1, 0.95]`

Si calculamos el producto punto entre el Vector A y el Vector B:
$$ A \cdot B = (0.9 \times 0.8) + (0.1 \times 0.0) = 0.72 + 0 = 0.72 $$

Si calculamos el producto punto entre el Vector A y el Vector C:
$$ A \cdot C = (0.9 \times 0.1) + (0.1 \times 0.95) = 0.09 + 0.095 = 0.185 $$

### El mecanismo de amplificación
El producto punto funciona tan bien para encontrar similitud porque actúa como un filtro multiplicador. Solo la puntuación crece de forma significativa cuando *ambos* vectores tienen números altos en la *misma* dimensión. Si un vector tiene un valor alto y el otro tiene un **0** (como ocurre en la dimensión de Finanzas entre los vectores A y B), la multiplicación se anula, ignorando inteligentemente esa característica no compartida.

### El truco en la Arquitectura de Software
En plataformas de búsqueda y relevancia complejas, o cuando se orquesta la memoria para agentes autónomos procesando miles de tokens de contexto, realizar las divisiones y las raíces cuadradas del denominador en la fórmula completa de la Similitud del Coseno para millones de vectores es un cuello de botella computacional inmenso.

La solución matemática que se aplica en la industria es **normalizar** todos los embeddings en la base de datos desde el principio. Al normalizar, fuerzas matemáticamente a que la longitud de todos los vectores sea exactamente **1** ($\|A\| = 1$). 

Si $\|A\| = 1$ y $\|B\| = 1$, el denominador de la fórmula del coseno desaparece ($1 \times 1 = 1$), dejando la ecuación reducida a:

$$ \cos(\theta) = A \cdot B $$

Por esta razón, el producto punto es el núcleo de las comparaciones de vectores. Permite calcular la similitud semántica exacta a una velocidad vertiginosa utilizando exclusivamente sumas y multiplicaciones, habilitando búsquedas de información y procesamiento de IA con latencias de milisegundos.
